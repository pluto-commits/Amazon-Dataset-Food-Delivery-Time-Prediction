```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/food_delivery_clean.csv")
```

```python
#What is the distribution of delivery time?
plt.figure(figsize=(8,5))

sns.histplot(df["Time_taken(min)"], bins=20, kde=True)

plt.title("Distribution of Delivery Time")
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Deliveries")

plt.show()
```

Most deliveries are completed between 20–35 minutes, with an average delivery time of approximately 26.5 minutes. The distribution appears moderately right-skewed, indicating that while most deliveries are completed within the expected time, a smaller number experience longer delays.

```python
# Does driver age affect delivery time?

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Delivery_person_Age",
    y="Time_taken(min)"
)

plt.title("Driver Age vs Delivery Time")

plt.show()

df[["Delivery_person_Age", "Time_taken(min)"]].corr()
```

Driver age has little linear relationship with delivery time.

```python
# Do higher rated drivers deliver faster?

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Delivery_person_Ratings",
    y="Time_taken(min)"
)

plt.xticks(rotation=45)

plt.show()
```

Better-rated delivery partners tend to complete deliveries more efficiently.

```python
# Does the number of multiple deliveries increase delivery time?

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="multiple_deliveries",
    y="Time_taken(min)"
)

plt.show()
```

```python
# Relationship between restaurant and delivery locations

plt.figure(figsize=(8,6))

plt.scatter(
    df["Restaurant_longitude"],
    df["Restaurant_latitude"],
    alpha=0.4,
    label="Restaurant"
)

plt.scatter(
    df["Delivery_location_longitude"],
    df["Delivery_location_latitude"],
    alpha=0.4,
    label="Delivery"
)

plt.legend()

plt.title("Restaurant and Delivery Locations")

plt.show()
```

```python
# Correlation between numerical variables

plt.figure(figsize=(12,8))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.show()
```

```python
# Distribution of Driver ratings

plt.figure(figsize=(8,5))

sns.histplot(
    df["Delivery_person_Ratings"],
    bins=20,
    kde=True
)

plt.show()
```

```python
# Age Distribution

plt.figure(figsize=(8,5))

sns.histplot(
    df["Delivery_person_Age"],
    bins=20,
    kde=True
)

plt.show()
```
