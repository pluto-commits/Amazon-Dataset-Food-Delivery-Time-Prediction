```python
import pandas as pd

# Load your dataset
df = pd.read_csv("../data/updated.csv")
```

```python
df_clean = df.copy()
df_clean.drop(columns=["Unnamed: 0"], inplace = True)
df_clean.head()
```

```python
print("Before:", df_clean.shape)
df_clean.drop_duplicates(inplace = True)
print("After:", df_clean.shape)
```

```python
df_clean[df_clean["Delivery_person_Age"]==0]

median_age = df_clean.loc[
    df_clean["Delivery_person_Age"] > 0,
    "Delivery_person_Age"
].median()

df_clean.loc[
    df_clean["Delivery_person_Age"] == 0,
    "Delivery_person_Age"
] = median_age

df_clean["Delivery_person_Age"].describe()
```

```python
import numpy as np

df_clean.loc[
    df_clean["Delivery_person_Ratings"] > 5,
    "Delivery_person_Ratings"
] = np.nan
```

```python
median_rating = df_clean["Delivery_person_Ratings"].median()

df_clean["Delivery_person_Ratings"].fillna(
    median_rating,
    inplace=True
)
```

```python
df_clean["multiple_deliveries"].value_counts(dropna=False)

mode_delivery = df_clean["multiple_deliveries"].mode()[0]

df_clean["multiple_deliveries"].fillna(
    mode_delivery,
    inplace=True
)
```

```python
df_clean[df_clean["Time_Orderd"].isnull()]
```

```python
df_clean = df_clean.dropna(subset=["Time_Orderd"])
```

```python
df_clean.isnull().sum()
```

```python
print(df_clean["Delivery_person_Ratings"].isna().sum())
print(df_clean["multiple_deliveries"].isna().sum())
```

```python
print(df_clean["Delivery_person_Ratings"].unique())
print(df_clean["multiple_deliveries"].unique())
```

```python
median_rating = df_clean["Delivery_person_Ratings"].median()

df_clean["Delivery_person_Ratings"] = (
    df_clean["Delivery_person_Ratings"]
    .fillna(median_rating)
)
```

```python
df_clean["Delivery_person_Ratings"].isna().sum()
```

```python
mode_delivery = df_clean["multiple_deliveries"].mode()[0]

df_clean["multiple_deliveries"] = (
    df_clean["multiple_deliveries"]
    .fillna(mode_delivery)
)
```

```python
df_clean["multiple_deliveries"].isna().sum()
```

```python
df_clean.isnull().sum()
```
